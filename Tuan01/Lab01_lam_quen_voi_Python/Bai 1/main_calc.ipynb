## Bài làm

### 1. Xem nội dung file `calc.py`

```python
print(open('calc.py', encoding='utf-8').read())
```

### 2. Chạy trực tiếp file `calc.py` bằng `%run`

```python
%run calc.py
```

### 3. Gọi lại các hàm vừa được nạp từ `calc.py`

```python
print('add(3, 4)      =', add(3, 4))
print('sub(3, 4)      =', sub(3, 4))
print('mul(3, 4)      =', mul(3, 4))
print('div(3, 4)      =', div(3, 4))
print('power(2, 10)   =', power(2, 10))
print('sqrt(2)        =', sqrt(2))
```

### 4. Dùng `calc.py` như một module

```python
import importlib
import calc

importlib.reload(calc)  # nạp lại khi file calc.py có thay đổi

print('calc.add(10, 5) =', calc.add(10, 5))
print('calc.div(10, 5) =', calc.div(10, 5))
print('calculate("10 * 5") =', calc.calculate('10 * 5'))
```

### 5. Thử nghiệm hàm `calculate` với biểu thức chuỗi

```python
exprs = ['1 + 2 * 3', '(4 + 6) / 5', '2 ** 10', '7 % 3', '10 / 0']

for e in exprs:
    try:
        print(f'{e:>12s} = {calc.calculate(e)}')
    except Exception as err:
        print(f'{e:>12s} -> Lỗi: {err}')
```

### 6. Kiểm thử nhanh (unit test) các hàm của `calc.py`

```python
assert calc.add(2, 3) == 5
assert calc.sub(2, 3) == -1
assert calc.mul(2, 3) == 6
assert calc.div(6, 3) == 2
assert calc.power(2, 5) == 32
assert abs(calc.sqrt(9) - 3) < 1e-9

try:
    calc.div(1, 0)
except ZeroDivisionError as err:
    print('Bắt đúng lỗi chia 0:', err)

print('=> Tất cả test đều PASS')
```

### 7. Chạy `calc.py` bằng lệnh shell

Ngoài `%run`, có thể gọi trình thông dịch Python qua magic `!` của notebook.

```python
import os
import subprocess
import sys

env = os.environ.copy()
env['PYTHONIOENCODING'] = 'utf-8'  # ép tiến trình con xuất UTF-8

result = subprocess.run(
    [sys.executable, '-X', 'utf8', 'calc.py'],
    capture_output=True, text=True, encoding='utf-8', env=env,
)

print(result.stdout)
if result.stderr:
    print('STDERR:', result.stderr)
print('return code =', result.returncode)
```
